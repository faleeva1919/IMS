```python
import pandas as pd
import numpy as np




# Таблица продаж
sales = pd.DataFrame({
    'OrderID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Product': ['Laptop', 'Phone', 'Tablet', 'Laptop', 'Phone',
                'Laptop', 'Tablet', 'Phone', 'Laptop', 'Tablet'],
    'Quantity': [1, 2, 1, 3, 1, 2, 1, 4, 1, 2],
    'Price': [50000, 30000, 20000, 50000, np.nan,
              50000, 20000, 30000, 50000, np.nan],
    'Region': ['Moscow', 'SPb', 'Kazan', 'Moscow', 'SPb',
               'Kazan', np.nan, 'SPb', 'Moscow', 'Kazan']
})


# Таблица товаров
products = pd.DataFrame({
    'Product': ['Laptop', 'Phone', 'Tablet', 'Monitor'],
    'Category': ['Electronics', 'Electronics', 'Electronics', 'Accessories'],
    'Weight_kg': [2.5, 0.2, 0.5, 3.0]
})


print("Таблица продаж:")
print(sales)
print("\nТаблица товаров:")
print(products)
```

```python
#Вариант 5
print("\nЗадание 1. Среднее значение Quantity для каждого продукта:")
print(sales.groupby('Product')['Quantity'].mean())
```

```python
sales['Revenue'] = sales['Quantity'] * sales['Price']

total_revenue = sales.groupby('Product')['Revenue'].sum()

print("\nЗадание 2. Продукт с максимальной общей выручкой:")
print(total_revenue.idxmax())
print("Максимальная общая выручка:", total_revenue.max())
```

```python
sales['Region'] = sales['Region'].ffill()

print("\nЗадание 3. Region после заполнения пропусков методом forward fill:")
print(sales['Region'])
```

```python
print("\nЗадание 4. Данные, отсортированные по OrderID в обратном порядке:")
print(sales.sort_values(by='OrderID', ascending=False))
```

```python
print("\nЗадание 5. Результат outer join таблиц sales и products:")
print(pd.merge(sales, products, on='Product', how='outer'))
```
