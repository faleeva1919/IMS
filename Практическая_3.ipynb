```python
import pandas as pd
import numpy as np

order_files = [
    "orders_2024_01.csv",
    "orders_2024_02.csv",
    "orders_2024_03.csv"
]

df_orders = pd.concat(
    [pd.read_csv(file, parse_dates=["order_datetime"]) for file in order_files],
    ignore_index=True
)

sheets = pd.read_excel(
    "products.xlsx",
    sheet_name=["products", "categories"]
)

df_products = sheets["products"]
df_categories = sheets["categories"]

print("Таблица заказов:")
print(df_orders.head())

print("\nТаблица товаров:")
print(df_products.head())

print("\nТаблица категорий:")
print(df_categories.head())

print("\nИнформация о заказах:")
df_orders.info()

print("\nИнформация о товарах:")
df_products.info()

print("\nИнформация о категориях:")
df_categories.info()
```

```python
df_prod_full = df_products.merge(
    df_categories,
    on="category_id",
    how="left",
    suffixes=("", "_cat")
)

print("\nЗадание 2. Расширенный справочник товаров:")
print(df_prod_full.head())
```

```python
df_orders_prod = df_orders.merge(
    df_prod_full,
    on="product_id",
    how="left"
)

print("\nЗадание 3. Объединенная таблица заказов и товаров:")
print(df_orders_prod.head())
```

```python
df_orders_prod["Revenue"] = df_orders_prod["price"] * df_orders_prod["quantity"]

category_revenue = df_orders_prod.groupby("category_name")["Revenue"].sum()

print("\nЗадание 4. Выручка по категориям товаров:")
print(category_revenue)
```

```python
top_categories = category_revenue.sort_values(ascending=False).head(3)

print("\nЗадание 5. Топ-3 категории по выручке:")
print(top_categories)

top_categories.to_csv("top_categories.csv", index=True)
```
