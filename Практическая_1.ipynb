```python
import pandas as pd
import numpy as np

data = {
    'ID': range(1, 11),
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva',
             'Frank', 'Grace', 'Helen', 'Ivan', 'Julia'],
    'Age': [23, 35, 29, 40, 28, 45, 32, 27, 38, 31],
    'City': ['Moscow', 'SPb', 'Moscow', 'Kazan', 'SPb',
             'Moscow', 'SPb', 'Kazan', 'Moscow', 'SPb'],
    'Salary': [50000, 70000, 60000, 80000, 55000,
               90000, 65000, 58000, 75000, 62000]
}

df = pd.DataFrame(data)

print("Исходный DataFrame:")
print(df)
```

```python
#Вариант 5
print("\nЗадание 1. Размер DataFrame:")
print(df.shape)
```

```python
print("\nЗадание 2. Первые 5 строк и столбцы Name, Age:")
print(df.loc[:4, ['Name', 'Age']])
```

```python
print("\nЗадание 3. Жители Kazan или SPb:")
print(df[df['City'].isin(['Kazan', 'SPb'])])
```

```python
print("\nЗадание 4. Минимальная зарплата:")
print(df['Salary'].min())
```

```python
df['Bonus'] = df['Salary'] * 0.10

print("\nЗадание 5. DataFrame с новым столбцом Bonus:")
print(df)
```
